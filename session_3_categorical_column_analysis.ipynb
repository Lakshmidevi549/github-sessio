```python
visa_df#df
visa_df['continent'] #one column
con=visa_df['continent']=='Asia' #one class
np.sum(con) #sum
visa_df[con] #extract
```

```python
cont_class=visa_df['continent'].unique()
freq=[]
for i in cont_class:
    con=visa_df['continent']==i
    freq.append(np.sum(con))
pd.DataFrame(zip(cont_class,freq),
             columns=['class','frequence'])
```

```python
keys=visa_df['continent'].value_counts().keys().tolist()
values=visa_df['continent'].value_counts().values.tolist()
pd.DataFrame(zip(keys,values),
             columns=['class','frequence'])
```
